Outliers:

An outlier is a data point that sits far away from the bulk of the other values in a column. It is unusually high or unusually low compared to the rest.

Why they appear:

Errors: typos, sensor faults, or bad data entry (for example, age = 150 or credit_score = 0).

Rare but real values: a person with a very high income or a very large loan. These are valid, just uncommon.

How they are detected:

IQR rule (box plot): a value is an outlier if it is below Q1 - 1.5 × IQR or above Q3 + 1.5 × IQR, where IQR = Q3 - Q1.

Z-score: a value more than 3 standard deviations from the mean (works best on bell-shaped data).

Visual: dots beyond the whiskers on a box plot, or isolated bars far from the main mass on a histogram.

Data visualization for the detection of outliers:

Always use box plot to detect the outliers in the dataset.

```python
# Numeric columns only (drop the target if it's 0/1):
num_cols = df.select_dtypes(include="number").columns.drop(
    "loan_paid_back", errors="ignore"
)

# Grid of box plots:
n_cols = 4
n_rows = -(-len(num_cols) // n_cols)  # ceiling division
fig, axes = plt.subplots(n_rows, n_cols, figsize=(18, 4 * n_rows))
axes = axes.flatten()

for ax, col in zip(axes, num_cols):
    sns.boxplot(
        y=df[col], ax=ax, color="skyblue", flierprops={"marker": "o", "markersize": 3}
    )
    ax.set_title(col)
    ax.set_ylabel("")

# Hide any unused subplots:
for ax in axes[len(num_cols) :]:
    ax.axis("off")

plt.tight_layout()
plt.show()
```

Also use distribution plots to detect the outliers:

```python
num_cols = df.select_dtypes(include="number").columns.drop(
    "loan_paid_back", errors="ignore"
)

n_cols = 4
n_rows = -(-len(num_cols) // n_cols)
fig, axes = plt.subplots(n_rows, n_cols, figsize=(18, 4 * n_rows))
axes = axes.flatten()

for ax, col in zip(axes, num_cols):
    sns.histplot(df[col], kde=True, bins=50, ax=ax, color="steelblue")
    ax.axvline(df[col].mean(), color="red", linestyle="--", label="mean")
    ax.axvline(df[col].median(), color="green", linestyle="-", label="median")
    ax.set_title(f"{col} (skew={df[col].skew():.2f})")
    ax.set_xlabel("")
    ax.legend(fontsize=8)

for ax in axes[len(num_cols) :]:
    ax.axis("off")

plt.tight_layout()
plt.show()
```

Removal of outliers:

Method 02:

Remove rows using the z-score:

Import Libraries:

```python
import numpy as np
from scipy import stats
```

```python
cols = [
    "age",
    "annual_income",
    "monthly_income",
    "debt_to_income_ratio",
    "credit_score",
    "loan_amount",
    "interest_rate",
    "installment",
    "total_credit_limit",
    "current_balance",
]
z = np.abs(stats.zscore(df[cols], nan_policy="omit"))
df_clean = df[(z < 3).all(axis=1)]
```

```python
print(
    f"Rows before: {len(df)}, after: {len(df_clean)} "
    f"({1 - len(df_clean) / len(df):.1%} removed)"
)
```

Data visulation after droppping rows:

```python
# Before vs after distributions (histograms overlaid):

sns.set_style("whitegrid")

n = len(cols)
ncols = 3
nrows = -(-n // ncols)
fig, axes = plt.subplots(nrows, ncols, figsize=(15, 3.5 * nrows))
for ax, c in zip(axes.ravel(), cols):
    sns.histplot(
        df[c], color="tab:red", alpha=0.4, label="Before", stat="density", ax=ax
    )
    sns.histplot(
        df_clean[c], color="tab:blue", alpha=0.5, label="After", stat="density", ax=ax
    )
    ax.set_title(c)
    ax.set_xlabel("")
axes.ravel()[0].legend()
for ax in axes.ravel()[n:]:
    ax.axis("off")
plt.suptitle("Distributions: Before vs After Outlier Removal", y=1.01, fontsize=14)
plt.tight_layout()
plt.show()
```

```python
# Boxplots before vs after:

fig, axes = plt.subplots(nrows, ncols, figsize=(15, 3.5 * nrows))
for ax, c in zip(axes.ravel(), cols):
    sns.boxplot(data=[df[c], df_clean[c]], palette=["tab:red", "tab:blue"], ax=ax)
    ax.set_xticklabels(["Before", "After"])
    ax.set_title(c)
for ax in axes.ravel()[n:]:
    ax.axis("off")
plt.tight_layout()
plt.show()
```

```python
# Rows removed per column:

removed = {}
for c in cols:
    q1, q3 = df[c].quantile([0.25, 0.75])
    i = q3 - q1
    removed[c] = (~df[c].between(q1 - 1.5 * i, q3 + 1.5 * i)).mean() * 100
pd.Series(removed).sort_values().plot(kind="barh", figsize=(8, 5), color="tab:orange")
plt.xlabel("% of rows flagged as outliers")
plt.title("Outlier share per column")
plt.show()
```

```python
# Did the target balance change?

fig, axes = plt.subplots(1, 2, figsize=(8, 3.5))
for ax, d, t in zip(axes, [df, df_clean], ["Before", "After"]):
    d["loan_paid_back"].value_counts(normalize=True).sort_index().plot(
        kind="bar", ax=ax, color=["tab:red", "tab:green"]
    )
    ax.set_title(f"loan_paid_back ({t})")
    ax.set_ylabel("Proportion")
plt.tight_layout()
plt.show()
```

```python
# Rows before/after:

plt.bar(["Before", "After"], [len(df), len(df_clean)], color=["tab:red", "tab:blue"])
plt.title("Row count")
plt.show()
```
