Feature Selection:

There are the two techniques for the feature selection:

1) Backward Elimination.

2) Forward Selection.

Both are wrapper methods for feature selection. They repeatedly fit a model and add or remove features based on a criterion (p-value, AIC/BIC, adjusted R², or cross-validated score). What you called “forward elimination” is usually named forward selection.

Backward Elimination:

Start with all features in the model.

Choose a significance level (e.g., SL = 0.05).

Fit the model using all features and calculate the p-value for each feature.

Identify the feature with the highest p-value.

If its p-value is greater than SL (0.05), remove that feature from the model.

Refit the model using the remaining features and recalculate their p-values.

Repeat the process until all remaining features have p-values less than SL (0.05).

Stop when no remaining feature is insignificant.

```python
full_model = sm.Logit(y_train, sm.add_constant(X_train)).fit(disp=0)
pvals = full_model.pvalues.drop("const").sort_values(ascending=False)   # ignore the intercept.
```

```python
print("Top 8 WORST features in the full model (bigger p-value = worse):\n")
print(pvals.head(8).apply(lambda p: f"{p:.3f}"))
```

The full Backward Elimination function:

```python
def backward_elimination(X, y, threshold_out=0.05):

    features = list(X.columns)     # start with ALL features.
    step = 0

    while len(features) > 0:
        # Fit the model with the features we still have:
        model = sm.Logit(y, sm.add_constant(X[features])).fit(disp=0)

        # Find the feature with the LARGEST p-value (ignore the intercept "const"):
        p_values = model.pvalues.drop("const")
        worst_feature = p_values.idxmax()
        worst_p = p_values.max()

        if worst_p > threshold_out:           # too weak -> remove it.
            step += 1
            features.remove(worst_feature)
            print(f"Step {step:>2}: DROP {worst_feature:<32} p-value = {worst_p:.3f}")
        else:                                 # all remaining are good -> stop.
            print(f"\nStop: the worst remaining feature '{worst_feature}' has p = {worst_p:.2e} (not > {threshold_out})")
            break

    return features
```

```python
backward_features = backward_elimination(X_train, y_train, threshold_out=0.05)
print("\n Backward Elimination kept", len(backward_features), "features:")
for f in backward_features:
    print("   -", f)
```

Backward feature selection by Scikit-Learn:

Safety check (versions and earlier results):

```python
import time
import sklearn
from sklearn.feature_selection import SequentialFeatureSelector   # needs scikit-learn 0.24 or newer.
```

```python
print("scikit-learn version:", sklearn.__version__)
```

```python
print("Everything from the earlier parts is available.")
```

```python
k = len(backward_features)                       # same number of features as before.
base_model = LogisticRegression(max_iter=1000)

def run_sfs(direction):
    start = time.time()
    selector = SequentialFeatureSelector(
        base_model,
        n_features_to_select=k,
        direction=direction,        # "forward" or "backward".
        scoring="roc_auc",
        cv=5,
    )
    selector.fit(X_train, y_train)
    chosen = list(X_train.columns[selector.get_support()])
    print(f"{direction.capitalize():<9}: picked {len(chosen)} features in {time.time() - start:.0f} seconds")
    return chosen
```

Run backward:

```python
sk_backward = run_sfs("backward")
print(sorted(sk_backward))
```
