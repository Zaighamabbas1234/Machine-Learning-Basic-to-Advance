Replace missing values:

General Method:

```python
df.fillna(10)
```

This will replace all the missing values irrespective of their data types.

This will replace all the numeric and categorical values present in the dataset.

Example: NAN is replaced with the 10 in the InternetService.

Wrong way to replace the values.

Correct way to replace categorical values:

Categorical values simply filled the backward filling, forward filling and mode operation.

Using the backward filling:

```python
df.bfill(inplace=True)
```

```python
df.bfill(inplace=True, axis=0)
```

```python
df.bfill(inplace=True, limit= 1)
```

Data visualization:

```python
sns.heatmap(df.isnull())
plt.show()
```

Using the forward filling:

```python
df.ffill(inplace=True)
```

```python
df.ffill(inplace=True, axis= 0)
```

```python
df.ffill(inplace=True, limit= 1)
```

Data visualization:

```python
sns.heatmap(df.isnull())
plt.show()
```

By using the mode:

Most freqent occuring value in a column.

Column wise operation.

```python
df["InternetService"].mode()
```

```python
df["InternetService"].fillna(df["InternetService"].mode())
```

```python
df["InternetService"].fillna(df["InternetService"].mode()[0])
```

```python
df["InternetService"].fillna(df["InternetService"].mode()[0],inplace= True)
```

Data Visualization:

```python
sns.heatmap(df.isnull())
plt.show()
```
