Data visualization for the detection of outliers:

```python
num_cols = df.select_dtypes(include="number").columns.drop(
    "loan_paid_back", errors="ignore"
)

n_cols = 4
n_rows = -(-len(num_cols) // n_cols)
fig, axes = plt.subplots(n_rows, n_cols, figsize=(18, 4 * n_rows))
axes = axes.flatten()

for ax, col in zip(axes, num_cols):
    sns.histplot(df[col], kde=True, bins=50, ax=ax, color="steelblue")
    ax.axvline(df[col].mean(), color="red", linestyle="--", label="mean")
    ax.axvline(df[col].median(), color="green", linestyle="-", label="median")
    ax.set_title(f"{col} (skew={df[col].skew():.2f})")
    ax.set_xlabel("")
    ax.legend(fontsize=8)

for ax in axes[len(num_cols) :]:
    ax.axis("off")

plt.tight_layout()
plt.show()
```

Removal of outliers:

```python
def remove_outliers_iqr(data, cols, k=1.5):
    mask = pd.Series(True, index=data.index)
    for col in cols:
        q1, q3 = data[col].quantile([0.25, 0.75])
        iqr = q3 - q1
        if iqr == 0:  # skip near-constant columns
            continue
        mask &= data[col].between(q1 - k * iqr, q3 + k * iqr)
    return data[mask]


# Numeric, continuous columns only (no target, no categoricals, no sparse counts):
cols = [
    "age",
    "annual_income",
    "monthly_income",
    "debt_to_income_ratio",
    "credit_score",
    "loan_amount",
    "interest_rate",
    "installment",
    "total_credit_limit",
    "current_balance",
]

df_clean = remove_outliers_iqr(df, cols)
```

```python
print(
    f"Rows before: {len(df)}, after: {len(df_clean)} "
    f"({1 - len(df_clean) / len(df):.1%} removed)"
)
```

Data visulation after droppping rows:

```python
# Before vs after distributions (histograms overlaid):

sns.set_style("whitegrid")

n = len(cols)
ncols = 3
nrows = -(-n // ncols)
fig, axes = plt.subplots(nrows, ncols, figsize=(15, 3.5 * nrows))
for ax, c in zip(axes.ravel(), cols):
    sns.histplot(
        df[c], color="tab:red", alpha=0.4, label="Before", stat="density", ax=ax
    )
    sns.histplot(
        df_clean[c], color="tab:blue", alpha=0.5, label="After", stat="density", ax=ax
    )
    ax.set_title(c)
    ax.set_xlabel("")
axes.ravel()[0].legend()
for ax in axes.ravel()[n:]:
    ax.axis("off")
plt.suptitle("Distributions: Before vs After Outlier Removal", y=1.01, fontsize=14)
plt.tight_layout()
plt.show()
```

Function transformation:

What is skewness?

Skewness is a single number that tells us the shape:

Near (0) = balanced (like a bell curve).

Positive and big =  long tail on the right (a few huge values).

Negative and big = long tail on the left.

Helper function:

```python
def compare(before, after, title_after):
    fig, axes = plt.subplots(1, 2, figsize=(12, 3.8))
    sns.histplot(before, bins=40, ax=axes[0], color="tomato")
    axes[0].set_title(f"Before  (skew = {before.skew():.2f})")
    sns.histplot(after, bins=40, ax=axes[1], color="seagreen")
    axes[1].set_title(f"After {title_after}  (skew = {after.skew():.2f})")
    for ax in axes:
        ax.set_xlabel("")
    plt.tight_layout()
    plt.show()
```

Import necessary library:

```python
from sklearn.preprocessing import PowerTransformer
```

Box-Cox transformation:

```python
bc = PowerTransformer(method="box-cox")
```

```python
df["limit_boxcox"] = bc.fit_transform(df[["total_credit_limit"]]).ravel()
```

```python
print("Chosen lambda:", round(bc.lambdas_[0], 3))
```

Data visualization:

```python
compare(df["total_credit_limit"], df["limit_boxcox"], "Box-Cox")
```

Yeo-Johnson transformation:

```python
yj = PowerTransformer(method="yeo-johnson")
```

```python
df["limit_yj"] = yj.fit_transform(df[["total_credit_limit"]]).ravel()
```

Data visualization:

```python
compare(df["total_credit_limit"], df["limit_yj"], "Yeo-Johnson")
```

Compare all methods side by side:

```python
plots = {
    # Row 1: annual_income:
    "Original income": "annual_income",
    "Log": "income_log",
    "Square root": "income_sqrt",
    "Reciprocal": "income_recip",
    # Row 2: total_credit_limit:
    "Original credit limit": "total_credit_limit",
    "Cube root": "limit_cbrt",
    "Box-Cox": "limit_boxcox",
    "Yeo-Johnson": "limit_yj",
}

fig, axes = plt.subplots(2, 4, figsize=(18, 7))

for ax, (title, col) in zip(axes.flat, plots.items()):
    sns.histplot(
        df[col],
        bins=40,
        kde=True,
        ax=ax,
        color="tomato" if title.startswith("Original") else "steelblue",
    )
    ax.set_title(f"{title}\nskew = {df[col].skew():.2f}")
    ax.set_xlabel("")
    ax.set_ylabel("")

plt.suptitle("Effect of Different Transformations on Skewness", fontsize=14, y=1.02)
plt.tight_layout()
plt.show()
```

```python
summary = pd.DataFrame(
    {
        "Column": ["annual_income"] * 4 + ["total_credit_limit"] * 4,
        "Method": list(plots.keys()),
        "Skewness": [round(df[col].skew(), 2) for col in plots.values()],
    }
)

# Rank within each column: closest to 0 is best:
summary["Abs skew"] = summary["Skewness"].abs()
summary["Best"] = summary["Abs skew"] == summary.groupby("Column")[
    "Abs skew"
].transform("min")
summary["Best"] = summary["Best"].map({True: "✅", False: ""})

summary = summary.drop(columns="Abs skew")
summary
```
