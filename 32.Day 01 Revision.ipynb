Day 01 Revision: Days 1 to 32 of Data Preprocessing:

Before moving on to models, I stopped to revise everything I’ve learned. Here is the full preprocessing journey so far:

◆ Days 1-3: Exploring and Handling Missing Data: head(), tail(), describe(), info(), isnull(), heatmaps and deleting rows or columns with dropna() and drop().

◆ Days 4-13: Replacing Missing Values: fillna(), bfill(), ffill(), mode, mean and median imputation then scikit-learn’s SimpleImputer with mean, median, most_frequent and constant strategies.

◆ Days 14-16: Encoding Categorical Data: One-Hot, Label and Ordinal encoding and when each one fits.

◆ Days 17-19: Outliers: detecting them with box plots and histograms, then removing them with the IQR and Z-score methods.

◆ Days 20-21: Feature Scaling: Standardization (StandardScaler) and Normalization (MinMaxScaler).

◆ Days 22-24: Data Quality: removing duplicates, replacing values with .replace() and cutting memory with astype().

◆ Days 25-31: Fixing Skewed Data: log, square root, reciprocal, cube root, Box-Cox and Yeo-Johnson then comparing them all side by side.
