Import the required library:

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OrdinalEncoder
```

SimpleImputer:

```python
imputer = SimpleImputer(strategy="most_frequent")  # most_frequent = mode
```

```python
cate_cols = df.select_dtypes(include="object").columns
df[cate_cols] = imputer.fit_transform(df[cate_cols])
```

Data visualization:

```python
sns.heatmap(df.select_dtypes(include="object").isnull())
plt.show()
```

OrdinalEncoding:

```python
encoder = OrdinalEncoder()
```

```python
cols = ["gender", "marital_status", "education_level", "employment_status"]
encoded_data = encoder.fit_transform(df[cols])
```

Convert encoded data to dataFrame:

```python
encoded_df = pd.DataFrame(encoded_data, columns=cols, index=df.index)
```

Remove original categorical columns:

```python
remaining_df = df.drop(columns=cols)
```

Combining Dataset:

```python
final_df = pd.concat([remaining_df, encoded_df], axis=1)
```

Display the head() of the Dataset:

```python
print("First 5 Rows of the Dataset:")
display(final_df.head(5))
```

Data visualziation:

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

for ax, col in zip(axes.ravel(), cols):
    counts = final_df[col].value_counts().sort_index()

    counts.plot(kind="bar", ax=ax)

    ax.set_title(f"Distribution of Encoded {col}")
    ax.set_xlabel(col)
    ax.set_ylabel("Count")

plt.tight_layout()
plt.show()
```

Ordinal Encoding also done by mapping.
Always use the scikit_Learn library OrdinalEncoding.
