Duplicates Values:

Not every duplicate is an error.

Is it a true duplicate? 

The same record entered twice (e.g., a data-entry or merge error).

Is it a legitimate repeat? 

Two orders from the same customer, or two players with the same name, are valid rows.

Is a key column involved? 

Use an ID column (or a combination of columns) to define what “duplicate” means.

```python
df.duplicated() # count fully identical rows.
```

```python
df.duplicated().sum() # count fully identical rows.
```

```python
df.duplicated().sum().sum() # count fully identical rows.
```

```python
df[df.duplicated(keep="first")] # view all duplicated rows.
```

```python
df[df.duplicated(keep="last")] # view all duplicated rows.
```

```python
df[df.duplicated(keep=False)] # view all duplicated rows.
```

```python
df.duplicated(subset=["charges"]) # duplicates based on specific columns.
```

```python
df.duplicated(subset=["charges"]).sum() # duplicates based on specific columns.
```

```python
df.duplicated(subset=["charges"]).sum().sum() # duplicates based on specific columns.
```
